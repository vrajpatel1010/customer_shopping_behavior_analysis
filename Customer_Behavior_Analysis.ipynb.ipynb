```python
import pandas as pd

df=pd.read_csv("customer_shopping_behavior.csv")
```

```python
df["Review Rating"] = df.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ", "_", regex=False)
df = df.rename(columns={"purchase_amount_(usd)" : "purchase_amount"})
```

```python
df.columns
```

```python
#create a column age_group
labels = ['Young Adult','Adult','Middle-aged','Senior']
df['age_group'] = pd.qcut(df['age'], q=4, labels = labels)
```

```python
df[['age','age_group']].head(10)
```

```python
# create colums purchase_frequency_days

frequency_mapping = {'Fortnightly' : 14 , 'Weekly' : 7, 'Monthly' : 30, 'Quarterly' : 90, 'Bi-Weekly' : 14, 'Annually' : 365, 'Every 3 months' : 90}
df['purchase-frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase-frequency_days','frequency_of_purchases']].head(10)
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
df = df.drop('promo_code_used',axis=1)
```
